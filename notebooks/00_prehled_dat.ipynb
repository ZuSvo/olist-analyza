Dosavadní zjištění a úvahy:

B - Soubor Orders:Počty řádků s chybějícími daty:
order_approved_at                 160
order_delivered_carrier_date     1783
order_delivered_customer_date    2965
    approved_at: Hypotéza k ověření: proč nebyla objednávka schválena? 
        Metodou Crosstab se ukazuje, že datum schválení chybí u 141 "canceled" objednávek, které mohly být zrušeny krátce po vytvoření, 5 objednávek "created", které na schválení čekají a kupodivu 14 "delivered" orders, což by mohla být chyba v datech, 141+5+14=160
    delivered_carrier_date: Hypotéza k ověření: Jedná se "pouze" o kombinaci stavu před předáním přepravci nebo jsou zde i nějaké chyby v datech? 
        Crosstab: Approved 2 + Canceled 550 + Created 5 + Delivered 2 + Invoiced 314 + Processing 301 + Unuvailable 609 = 1783, tj. teoreticky chybou by mohly být 2 čádky stavu "delivered" - zde by mohla být lidská chyba zadání dat
    delivered_customer_date: Hypotéza k ověření: Jedná se opět pouze o kombinaci stavů nebo jsou zde i chyby v datech? Proč nebyly zásilky dodány?
        Crosstab: Approved 2 + Canceled 619 + Created 5 + Delivered 8 + Invoiced 314 + Processing 301 + Shipped 1107 + Unavailable 609, tj. diskutabilní jsou chybějící data u stavu Delivered
